```python
# Write logic to ask the user for their age & determine if they are eligible to vote 
# based on whether they are 18 or older.

age = int(input("Enter your age"))
if (age >= 18):
    print("You are eligible for Voting")
else:
    print("You are not eligible for Voting")
    
```

```python
# Write logic to identify and return the largest number from a given list.

numbers = [23, 12, 34, 29, 67, 49]
largest = numbers[0]
for num in numbers:
    if num > largest:
        largest = num
print("The largest number is:", largest)
```

```python
#  Write logic to determine the bonus amount based on the given salary.

salary = float(input("Enter your salary"))

if (salary >= 50000):
    bonus = salary * 0.10
else:
    bonus = 0
print("Bonus Amount:", bonus)
```

```python
# Write logic to check whether a given number is even or odd.

numCheck = int(input("Enter a number:"))

if (numCheck % 2 == 0):
   print("Even number")
else:
    print("Odd number")
    
```

```python
# Write logic to take a word or sentence as input and produce its reversed version.

anyWord = input("Enter a word or sentence to reverse:")
print("The Result is:", anyWord[::-1])

text =  input("Enter a sentence to reverse:")
word = text.split()
print(" ".join(word[::-1]))
```

```python
# Write logic to check if a student has passed a subject by scoring at least 40 marks.

score = int(input("Enter the student's score: "))
if score >= 40:
    print("PASS")
else:
    print("FAIL")
    
```

```python
# Write logic to calculate the final amount to be paid after applying the discount.

bill = float (input("Enter total order amount:"))
if bill > 1500:
    final_bill = bill - (bill * 0.20)
else:
    final_bill = bill
print("Amount to be paid:", final_bill)
```

```python
# Write logic to check if a user has enough balance before allowing a withdrawal and 
# update the remaining balance accordingly.

current_balance = 1000

withdraw = float(input("Enter withdrawal amount: "))
remaining_balance = current_balance - withdraw

if withdraw <= current_balance:
    print("Withdrawal successful.")
    print("Remaining balance:", remaining_balance)
else:
    print("Insufficient balance.")
```

```python
# Write logic to determine whether a given year is a leap year.

year = int(input("Enter year: "))
if (year % 400 == 0) or (year % 4 == 0 and year % 100 != 0):
    print(year, "is a Leap Year")
else:
    print(year, "is not a Leap Year")
    
```

```python
# Write logic to extract and return only the even numbers from a list.

list = [23, 45, 16, 34, 25, 7, 20, ]
even_numbers = []
for num in list:
    if (num % 2 == 0):
        even_numbers.append(num)
print("The Filtered Even Numbers are:", even_numbers)
```
